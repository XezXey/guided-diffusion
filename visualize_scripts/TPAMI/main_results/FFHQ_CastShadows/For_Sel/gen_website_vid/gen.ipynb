```python
import numpy as np
import glob, os
import json

n_frames = 360
ex = f'spiral_{n_frames}'
vid = f"/data/mint/sampling/TPAMI/main_result/ffhq/teaser/log=backup_paired+difareli+cs+nodpm+trainset_256_cfg=paired+difareli+cs+nodpm+trainset_256.yaml_spiral/ema_300000/valid/render_face/reverse_sampling/"
pf = f"/Lerp_1000/n_frames={n_frames}/"

js = '/home/mint/Dev/DiFaReli/difareli-faster/visualize_scripts/TPAMI/main_results/FFHQ_CastShadows/For_Sel/aj_ake_top_candidates.json'
with open(js, 'r') as f:
    data = json.load(f)['pair']
    
for pid, p in data.items():
    print(pid, p)
    src = p['src']
    dst = p['dst']
    path = f'{vid}/src={src}/dst={dst}/{pf}/'
    imgs = glob.glob(f'{path}/res_frame*.png')
    render = glob.glob(f'{path}/dst_ren_frame*.png')
    shadm = glob.glob(f'{path}/dst_shadm_shad_frame*.png')
    # Copy to current directory
    os.makedirs(f'./out/{ex}/src={src}/imgs/', exist_ok=True)
    for img in imgs:
        # Rename to 00000.png, 00001.png, ...
        if 'res_frame0' in img: continue
        idx = int(img.split('/')[-1].split('frame')[-1].split('.')[0])
        os.system(f'cp {img} ./out/{ex}/src={src}/imgs/{idx:05d}_res.png')
    for img in render:
        # Rename to 00000.png, 00001.png, ...
        if 'dst_ren_frame0' in img: continue
        idx = int(img.split('/')[-1].split('frame')[-1].split('.')[0])
        os.system(f'cp {img} ./out/{ex}/src={src}/imgs/{idx:05d}_ren.png')
    for img in shadm:
        # Rename to 00000.png, 00001.png, ...
        if 'dst_shadm_shad_frame0' in img: continue
        idx = int(img.split('/')[-1].split('frame')[-1].split('.')[0])
        os.system(f'cp {img} ./out/{ex}/src={src}/imgs/{idx:05d}_shadm.png')
        
    # Make video crf 17, 24fps
    os.system(f'ffmpeg -y -r 30 -i ./out/{ex}/src={src}/imgs/%05d_res.png -c:v libx264 -crf 17 -pix_fmt yuv420p ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_res.mp4')
    os.system(f'cp ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_res.mp4 ./out/{ex}/{pid}_src={src}_dst={dst}_res.mp4')
    os.system(f'ffmpeg -y -r 30 -i ./out/{ex}/src={src}/imgs/%05d_ren.png -c:v libx264 -crf 17 -pix_fmt yuv420p ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_ren.mp4')
    os.system(f'ffmpeg -y -r 30 -i ./out/{ex}/src={src}/imgs/%05d_shadm.png -c:v libx264 -crf 17 -pix_fmt yuv420p ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_shadm.mp4')
    
    # Horizontally concatenate
    os.system(f'ffmpeg -y -i ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_ren.mp4 -i ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_res.mp4 -i ./out/{ex}/src={src}/{pid}_src={src}_dst={dst}_shadm.mp4 -filter_complex "[0:v][1:v][2:v]hstack=inputs=3" -c:v libx264 -crf 17 -pix_fmt yuv420p ./out/{ex}/{pid}_src={src}_dst={dst}_all.mp4')
    
    
```
