# Check sampling progress on HDR

```python
import sys
from guided_diffusion.mint_logger import createLogger
logger = createLogger()

def print_progress(n, n_total, text, bar_length=30):
    progress = n / n_total
    block = int(bar_length * progress)
    bar = "█" * block + "-" * (bar_length - block)
    logger.info(f"Progress on {text}: |{bar}| {n}/{n_total} ({progress*100:.1f}%)")
```

```python
import os, glob, json

c_list = [1.0, 0.9, 0.8, 0.7, 0.6, 0.5]
sd_list = [75, 50, 25]

def proc(sample_json):
    with open(sample_json, 'r') as f:
        samples = json.load(f)['pair']
        
    hdr_list = ['064_hdrmaps_com_free_2K', '125_hdrmaps_com_free_2K', '117_hdrmaps_com_free_2K']
    for hdr_map in hdr_list:
        for c in c_list:
            for sd in sd_list:
                total_sj = len(samples)
                count = 0
                for pid, p in samples.items():
                    src = p['src']
                    dst = p['dst']
                    path = f"/data/mint/TPAMI_MajorRevision/Ours/ffhq_hdr_finale/log=paired+difareli+cs+nodpm+trainset_256_cfg=paired+difareli+cs+nodpm+trainset_256.yaml_SD{sd}_{c}C_sColor_Lmax10_rAxis1/ema_300000/valid/render_face_hdr/{hdr_map}/src={src}/dst={dst}/Lerp_1000/n_frames=60/"
                    if os.path.exists(path) and len(glob.glob(f"{path}/res_frame*.png")) == 60:
                        count += 1
                print_progress(count, total_sj, text=f"HDR {hdr_map}, c={c}, sd={sd}")
                    
sample_json = f'/home/mint/Dev/DiFaReli/difareli-faster/experiment_scripts/TPAMI/sample_json/TPAMI_MajorRevision/aj_ake_samples.json'
proc(sample_json)
sample_json = f'/home/mint/Dev/DiFaReli/difareli-faster/experiment_scripts/TPAMI/sample_json/TPAMI_MajorRevision/all_rotateSH.json'
proc(sample_json)
                    
        
```
