### 2. FFHQ_Target2 (n_frames=20)
### 3. FFHQ_Rotate (n_frames=60)

```python
# FFHQ_Target = ffhq_target (From TR) + ffhq_target2 (From ICCV samples)
```

```python
import numpy as np
import glob, os, sys
import json

path = '/data/mint/DPM_Dataset/Dataset_For_Baseline/'

def proc(path, out_dir, exclude, n_frames, sub_dat=[]):
    os.makedirs(out_dir, exist_ok=True)
    out_combined = {'pair':{}}
    new_pid = 0
    count = 0
    for sub in sub_dat:
        sub_pid = 0
        for i_p in sorted(glob.glob(f'{path}/{sub}/valid/*')):
            pair_id, src_id, dst_id = i_p.split('/')[-1].split('_')
            # print(pair_id, src_id, dst_id)
            if not os.path.exists(f'{i_p}/n_step={n_frames}'):
                continue
            out_combined['pair'][f'pair{new_pid}'] = {'src':src_id.split('=')[-1], 'dst':dst_id.split('=')[-1]}
            fp = f'{out_dir}/valid/pair{new_pid}_{src_id}_{dst_id}/n_step={n_frames}/'
            os.makedirs(f'{out_dir}/valid/pair{new_pid}_{src_id}_{dst_id}/n_step={n_frames}/', exist_ok=True)
            for i_f in glob.glob(f'{i_p}/n_step={n_frames}/*'):
                base_name = os.path.basename(i_f)
                if any(ex == base_name for ex in exclude):
                    continue
                else:
                    os.system(f'cp -r {i_f} {fp}')
            sub_pid += 1
            new_pid += 1

        print(f"[#] Total subjects for {sub}: {sub_pid}")
        count += sub_pid
    print(f"[#] Total subjects: {count} == {new_pid}")
    assert count == new_pid

    with open(f'{out_dir}/{out_dir.split("/")[-1]}.json', 'w') as f:
        json.dump(out_combined, f, indent=4)
    print(f"[#] Done!")
    print("#"*50)



proc(path, sub_dat=['ffhq_1'], exclude=['map', 'combined', 'map.mp4', 'combined.mp4'], n_frames=2, out_dir='/data/mint/DPM_Dataset/Dataset_For_Baseline/Dataset_For_Baseline_Submit/ffhq_target1')
proc(path, sub_dat=['ffhq_target', 'ffhq_target2'], exclude=['map', 'combined', 'map.mp4', 'combined.mp4'], n_frames=20, out_dir='/data/mint/DPM_Dataset/Dataset_For_Baseline/Dataset_For_Baseline_Submit/ffhq_target2')
proc(path, sub_dat=['ffhq_rotate', 'ffhq_shadows', 'ffhq_diffuse'], exclude=['map', 'combined', 'map.mp4', 'combined.mp4'], n_frames=60, out_dir='/data/mint/DPM_Dataset/Dataset_For_Baseline/Dataset_For_Baseline_Submit/ffhq_rotate')


```
