```python
import numpy as np
from PIL import Image
import cv2
import torchvision
import matplotlib.pyplot as plt
import torch as th
x = cv2.imread('./xdst.png.png')
print(x.shape)
print(np.unique(np.array(x)))
plt.imshow(x)
plt.show()

# Blur the image on various scales with for loops
out = []
# for i in np.linspace(1, 21, 2):
# for i in np.arange(1, 51, 2):
for sigma in np.linspace(0.1, 7, 60):
    sigma_temp = np.round(sigma)
    shape = sigma_temp*3*2 if (sigma_temp*3*2) % 2 == 1 else (sigma_temp*3*2)+1
    shape = int(shape)
    x_blurred = cv2.GaussianBlur(x.copy()/255., (shape, shape), sigma)
    out.append(x_blurred)

out = (np.array(out) * 255).astype(np.uint8)
print(out.shape)
torchvision.io.write_video('blurred.mp4', out, fps=20)

```
