# Gen grids from vids folder (After running gen_vids.py)

```python
import subprocess, os, glob, tqdm
ax = 1
videos = glob.glob(f"./vids/all_outputs/res/*_{ax}.mp4")
video_path = "./vids/all_outputs/res/"
# videos = glob.glob(f"./vids_rot1/all_outputs/res/*axis={ax}.mp4")

# Grid configuration
grid_h = 4  # Number of rows in the grid


sj = [
    [62848, 60208, 69757, 61234]
]

for i in range(len(sj)):
    for j in range(len(sj[i])):
        # path = glob.glob(f"{video_path}/src={sj[i][j]}.jpg_*_{ax}.mp4")
        path = glob.glob(f"{video_path}/{sj[i][j]}.jpg_*_{ax}.mp4")
        if len(path) == 0:
            print(f"Not found: {video_path}/src={sj[i][j]}.jpg_*_{ax}.mp4")
            continue
        sj[i][j] = path[0]
    print(sj[i])
    # Create grid video
    os.makedirs(f"./vids/all_outputs/grids_{ax}/", exist_ok=True)
    intermediate_files = []
    videos_to_use = sj[i]
    input_cmds = []
    for idx, video in enumerate(videos_to_use):
        input_cmds += ["-i", video]

    hstack_filter =  f'-filter_complex "vstack=inputs={len(videos_to_use)}"'
    output_grid = f"./vids/all_outputs/grids_{ax}/grid{i}.mp4"
    try:
        subprocess.run(f"ffmpeg {' '.join(input_cmds)} {hstack_filter} -y {output_grid}", shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL, check=True)
    except subprocess.CalledProcessError as e:
        print(e)
        exit(1)
```
