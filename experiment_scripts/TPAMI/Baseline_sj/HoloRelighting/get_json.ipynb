```python
pair_name_train = {'Fig4_main':{'11133.jpg':'f4_main_inp_1', '11734.jpg':'f4_main_inp_2', '27165.jpg':'f4_main_inp_3'},
            'Fig6_main':{'46881.jpg':'f6_main_inp_1', '46178.jpg':'f6_main_inp_2', '51645.jpg':'f6_main_inp_3'},
            'Fig6_supmat':{'27165.jpg':'f6_sup_inp_1', '11737.jpg':'f6_sup_inp_2', '19220.jpg':'f6_sup_inp_3', '27001.jpg':'f6_sup_inp_4', '27146.jpg':'f6_sup_inp_5', '11653.jpg':'f6_sup_inp_6', '11325.jpg':'f6_sup_inp_7'},
            'Fig7_supmat':{'6061.jpg':'f7_sup_inp_1', '46299.jpg':'f7_sup_inp_2', '67183.jpg':'f7_sup_inp_3', '51727.jpg':'f7_sup_inp_4', '6093.jpg':'f7_sup_inp_5'}
        }
pair_name_valid = {'Fig7_supmat':{'67183.jpg':'f7_sup_inp_3'}        
                }

train_pairs = {'pair':{}}
count = 0
for fig_name, pairs in pair_name_train.items():
    for img_name, pair_name in pairs.items():
        train_pairs['pair'][f'pair{count}'] = {'src':img_name, 'dst':pair_name.replace('inp', 'ref') + '.png', 'gt':pair_name.replace('inp', 'out') + '.png', 'inp':pair_name, 'fig_name':fig_name, 'pair_name':pair_name}
        count += 1

import json
with open('./holo_test_pairs.json', 'w') as f:
    json.dump(train_pairs, f, indent=4)
```

```python
# Make copy to the folder at /data/mint/DPM_Dataset/HoloRelighting/
import json, os
holo_path = './paper_sj/'
with open('./holo_test_pairs.json', 'r') as f:
    train_pairs = json.load(f)['pair'].items()

os.makedirs(f'/data/mint/DPM_Dataset/HoloRelighting/images/test/', exist_ok=True)
for k, v in train_pairs:
    if int(v['src'][:-4]) < 60000:
        data_path = '/data/mint/DPM_Dataset/ffhq_256_with_anno/ffhq_256/train/'
    else:
        data_path = '/data/mint/DPM_Dataset/ffhq_256_with_anno/ffhq_256/valid/'
    os.system(f"cp {data_path}/{v['src']} /data/mint/DPM_Dataset/HoloRelighting/images/test/")
    os.system(f"cp {holo_path}/{v['fig_name']}/{v['dst']} /data/mint/DPM_Dataset/HoloRelighting/images/test/")
    os.system(f"cp {holo_path}/{v['fig_name']}/{v['gt']} /data/mint/DPM_Dataset/HoloRelighting/images/test/")
```
