```python
import numpy as np
import pandas as pd
import json

in_path  = './targetSH_MJ/Batch_5346627_batch_results.csv'
out_path = './targetSH_MJ/Batch_5346627_batch_results_eiei.csv'

df = pd.read_csv(in_path)
df_out = df.copy()

count_total_q1 = 0
count_total_true_q1 = 0
count_total_q2 = 0
count_total_true_q2 = 0

for i in range(len(df_out)):
    raw = df_out.at[i, 'Answer.taskAnswers']
    try:
        # Expect a JSON list with one dict
        answers_list = json.loads(raw)
        if not isinstance(answers_list, list) or not answers_list:
            # Skip malformed rows
            continue
        answer = answers_list[0]
    except Exception:
        # Skip rows with bad JSON
        continue

    # We’ll build a new dict to avoid in-place iteration issues
    updated_answer = dict(answer)

    for k, v in answer.items():
        # k is like 'q1_a1_xxx' or 'q2_a2_xxx'; v is a dict {option_key: bool, ...}
        if "qnum" in k:
            continue
        try:
            q_idx = int(k.split('_')[0].replace('q', ''))
        except Exception:
            continue

        # Count totals per question index (move outside the diffusionrig branch if you want *all* questions)
        if q_idx % 2 == 0:
            count_total_q2 += 1
        else:
            count_total_q1 += 1

        if not isinstance(v, dict):
            continue

        option_keys = list(v.keys())
        if not option_keys:
            continue

        # Find diffusionrig option among keys
        dr_keys = [c for c in option_keys if 'diffusionrig' in c]
        if not dr_keys:
            continue

        foc = v
        k_foc = option_keys
        v_foc = [foc[k_] for k_ in k_foc]

        dr_key = dr_keys[0]
        dr_idx = k_foc.index(dr_key)
        dr_selected = bool(v_foc[dr_idx])

        if dr_selected:
            if q_idx % 2 == 1:  # Q1 (odd)
                count_total_true_q1 += 1
            else:               # Q2 (even)
                count_total_true_q2 += 1
                # Flip selection for Q2:
                # 1) turn off diffusionrig
                # 2) select the *other* option and label it as difareli++
                new_dict = {dr_key: False}
                # Assume binary choice; pick the other index
                other_idx = 1 - dr_idx if len(k_foc) == 2 else None
                if other_idx is not None:
                    other_key = k_foc[other_idx]
                    # Preserve original True/False structure but force True for the difareli++ choice
                    # Rename the other key to have 'difareli++' suffix (if that matches your schema)
                    # If you need exact names 'a1_difareli++' / 'a2_difareli++', reconstruct from index:
                    tag = f"a{other_idx+1}_difareli++"
                    new_dict[tag] = True
                else:
                    # Fallback: keep original map but turn off diffusionrig
                    new_dict.update({k_: (k_ != dr_key and foc[k_]) for k_ in k_foc})
                print("Before:", v)
                print(new_dict)
                updated_answer[k] = new_dict
    print(updated_answer)

    # Write back as JSON string (list with one dict, same as input)
    df_out.at[i, 'Answer.taskAnswers'] = json.dumps([updated_answer], ensure_ascii=False)

print("Total Q1: ", count_total_q1)
print("Total Q2: ", count_total_q2)
print("Total True Q1: ", count_total_true_q1)
print("Total True Q2: ", count_total_true_q2)

df_out.to_csv(out_path, index=False)
```
