```python
import numpy as np
import torch as th
from PIL import Image
import glob, os
from matplotlib import pyplot as plt
import tqdm

# dataset_name = 'ffhq_rotate'
# n_step = 60
# dataset_name = 'mp_test'
# dataset_name = 'mp_test2'
dataset_name = 'ffhq_target_figure'
n_step = 2
path = f'/data/mint/DPM_Dataset/Dataset_For_Baseline/{dataset_name}/' #valid/pair1000_src=66532.jpg_dst=61859.jpg/n_step=60'
outpath = f'/data/mint/DPM_Dataset/Dataset_For_Baseline/for_vis/{dataset_name}_vis_ball/' #valid/pair1000_src=66532.jpg_dst=61859.jpg/n_step=60'

fpath = f'{path}/valid/'
for p in tqdm.tqdm(glob.glob(fpath + 'pair*')):
    sj_path = f'{p}/n_step={n_step}/ball/'
    sj_name = p.split('/')[-1]

    os.makedirs(f'{outpath}/{sj_name}/n_step={n_step}/ball/' , exist_ok=True)
    for ball_p in glob.glob(sj_path + '*'):
        ball_img = Image.open(ball_p)
        # Make it transparent where the image is black
        ball_img = ball_img.convert("RGBA")
        ball_img_arr = np.array(ball_img)
        ball_img_arr[..., 3] = (ball_img_arr[..., :3] != 0).all(axis=-1) * 255
        ball_img = Image.fromarray(ball_img_arr)
        ball_img.save(f'{outpath}/{sj_name}/n_step={n_step}/ball/{ball_p.split("/")[-1]}')

        # plt.imshow(ball_img_arr)
        # plt.show()
    # assert False
```
