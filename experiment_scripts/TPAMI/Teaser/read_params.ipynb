```python
import numpy as np
import json

# light_path = np.load('./light_params.npy', allow_pickle=True)
light_traj = np.load('./light_traj.npy', allow_pickle=True).item()
light_traj, gen_params = light_traj['traj'], light_traj['params']
print(light_traj[0].keys())
radius = []
angle = []
rel_radius = []
rel_angle = []
for i in range(len(light_traj)):
    # print("[#] Angle: ", light_path[i]['angle'])
    # print("[#] Rel-Angle: ", light_path[i]['rel_angle'])
    # print("[#] Radius: ", light_path[i]['radius'])
    # print("[#] Rel-Radius: ", light_path[i]['rel_radius'])
    radius.append(light_traj[i]['radius'])
    rel_radius.append(light_traj[i]['rel_radius'])
    angle.append(light_traj[i]['angle'])
    rel_angle.append(light_traj[i]['rel_angle'])

import matplotlib.pyplot as plt
plt.plot(radius)
plt.title('Radius')
plt.show()

plt.plot(rel_radius)
plt.title('Relative Radius')
plt.show()

# plt.plot(np.cumsum([radius[0]] + rel_radius))
# plt.title('Cumulative Radius')
# plt.show()

plt.plot(angle)
plt.title('Angle')
plt.show()

plt.plot(rel_angle)
plt.title('Relative Angle')
plt.show()
```
