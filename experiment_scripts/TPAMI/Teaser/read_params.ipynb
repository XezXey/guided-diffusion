```python
import numpy as np
import json

light_path = np.load('./light_params.npy', allow_pickle=True)
print(light_path[0].keys())
radius = []
angle = []
rel_radius = []
rel_angle = []
for i in range(len(light_path)):
    # print("[#] Angle: ", light_path[i]['angle'])
    # print("[#] Rel-Angle: ", light_path[i]['rel_angle'])
    # print("[#] Radius: ", light_path[i]['radius'])
    # print("[#] Rel-Radius: ", light_path[i]['rel_radius'])
    radius.append(light_path[i]['radius'])
    rel_radius.append(light_path[i]['rel_radius'])
    angle.append(light_path[i]['angle'])
    rel_angle.append(light_path[i]['rel_angle'])

import matplotlib.pyplot as plt
plt.plot(radius)
plt.title('Radius')
plt.show()

plt.plot(rel_radius)
plt.title('Relative Radius')
plt.show()

# plt.plot(np.cumsum([radius[0]] + rel_radius))
# plt.title('Cumulative Radius')
# plt.show()

plt.plot(angle)
plt.title('Angle')
plt.show()

plt.plot(rel_angle)
plt.title('Relative Angle')
plt.show()
```
